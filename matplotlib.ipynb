```python
x = np.array([2, 8])
y = np.array([4, 10])

plt.plot(x, y, color='red')
plt.show()
```

```python
x=[2,4,6,8,10]
y=[3,5,3,8,15]
plt.figure(figsize=(9,5))
plt.plot(x,y,color='blue',linestyle=':',marker='o',linewidth=2)
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title("line chart")
plt.legend(['my data series'])
plt.grid(True)
plt.show()
```

```python
categories=['A','B','C','D','E']
values=[1,3,5,7,9]
plt.figure(figsize=(8,5))
plt.bar(categories,values,color='purple')
plt.xlabel('categories')
plt.ylabel('values')
plt.title('Bar chart')
plt.show()
```

```python
data = [10, 12, 14, 15, 16, 16, 19, 11, 14, 11, 14, 17, 18, 18, 15, 14]

# Create histogram
plt.hist(data, bins=20, edgecolor='purple')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.title('Simple Histogram Example')
plt.show()
```

```python
x = [3, 5, 7, 9, 11, 15, 18]
y = [1, 2, 3, 4, 5, 6, 7]

# Create scatter plot
plt.scatter(x, y, color='green')
plt.xlabel('X Value')
plt.ylabel('Y Value')
plt.title('Simple Scatter Plot Example')
plt.show()
```

```python
labels = ['A', 'B', 'C', 'D']
sizes = [18, 40, 65, 80]

# Create pie chart
plt.pie(sizes, labels=labels, autopct='%1.1f%%', startangle=120)
plt.title('Simple Pie Chart Example')
plt.axis('equal')  # Ensures the pie is drawn as a circle
plt.show()
```

```python
data = [4, 9, 6, 1, 12, 15, 19, 25, 7, 8,35 ]
# Create box plot
plt.boxplot(data)
plt.title('Simple Box Plot Example')
plt.ylabel('Values')
plt.show()
```

```python
from matplotlib.colors import LinearSegmentedColormap

# Define custom colors
colors = ["blue", "white", "red"]  # Blue → White → Red
custom_cmap = LinearSegmentedColormap.from_list("blue_white_red", colors)
data = np.random.rand(20, 20)
plt.imshow(data, cmap=custom_cmap, interpolation='nearest')
plt.colorbar()  # Show color scale
plt.title('Simple Heatmap Example')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-6, 6, 400)
y = np.linspace(-6, 6, 400)
X, Y = np.meshgrid(x, y)

R = np.sqrt(X**2 + Y**2)
Theta = np.arctan2(Y, X)

Z = np.sin(8 * Theta + R * 3)

plt.figure(figsize=(8, 8))
plt.imshow(Z, cmap='plasma', extent=(-6,6,-6,6))
plt.colorbar()
plt.title("Spiral Illusion Heatmap")
plt.axis('off')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-12, 12, 1200)
y = np.linspace(-12, 12, 1200)
X, Y = np.meshgrid(x, y)

R = np.sqrt(X**2 + Y**2)
A = np.arctan2(Y, X)

# Snake scales pattern
Z = np.sin(R*20 + 8*np.sin(A*6))
Z *= np.cos(A*15 + R)

plt.figure(figsize=(10,10))
plt.imshow(Z, cmap='turbo')
plt.axis('off')
plt.title("Snake Skin Vortex Illusion")

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-4, 4, 800)
y = np.linspace(-4, 4, 800)
X, Y = np.meshgrid(x, y)

Z = np.sin(X**2 * Y) + np.cos(Y**2 * X)

plt.figure(figsize=(8,8))
plt.imshow(Z, cmap='nipy_spectral')
plt.colorbar()
plt.title("Fractal Style")
plt.axis('off')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 100, 800)
y = np.linspace(0, 100, 800)
X, Y = np.meshgrid(x, y)

Z = np.sin(X) + np.sin(Y * 1.02)

plt.figure(figsize=(8,8))
plt.imshow(Z, cmap='inferno')
plt.colorbar()
plt.title("Moiré Illusion")
plt.axis('off')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-12, 12, 800)
y = np.linspace(-12, 12, 800)
X, Y = np.meshgrid(x, y)

R = np.sqrt(X**2 + Y**2)
T = np.arctan2(Y, X)

Z = np.sin(25 * T + 6 * R)

plt.figure(figsize=(8,8))
plt.imshow(Z, cmap='twilight')
plt.colorbar()
plt.title("Hypnotic Spiral")
plt.axis('off')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-10, 10, 600)
y = np.linspace(-10, 10, 600)
X, Y = np.meshgrid(x, y)

Z = np.sin(X * Y) + np.cos(X**2 - Y**2)

plt.figure(figsize=(8,8))
plt.imshow(Z, cmap='turbo')
plt.colorbar()
plt.title("Psychedelic Illusion")
plt.axis('off')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 20, 400)
y = np.linspace(0, 20, 400)
X, Y = np.meshgrid(x, y)

Z = np.sin(X * 5) * np.sin(Y * 5)

plt.figure(figsize=(8,8))
plt.imshow(Z, cmap='coolwarm')
plt.colorbar()
plt.title("Checkerboard Illusion")
plt.axis('off')
plt.show()
```
